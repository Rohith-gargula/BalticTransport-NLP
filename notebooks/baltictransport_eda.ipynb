# BalticTransport-NLP — EDA
This notebook is designed for Kaggle. The committed sample is synthetic smoke-test data; do not claim real-world findings from it.

```python
import pandas as pd
import matplotlib.pyplot as plt
DATA_PATH = '../data/sample_smoke_test.csv' # change to Kaggle input path after uploading reviewed dataset
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print('Rows:', len(df))
print('Languages:', df['language'].value_counts().to_dict())
print('Labels:', df['label'].value_counts().to_dict())
df.groupby(['language','label']).size().unstack(fill_value=0)
```

```python
df['word_count'] = df['text'].astype(str).str.split().str.len()
df.boxplot(column='word_count', by='language')
plt.title('Text length by language'); plt.suptitle(''); plt.show()
```
